#  Global pairwise Alignment using Biopython
The following part shows the alignement operation using Bio.Align

```python
# This cell imports Align module from Biopython
from Bio import Align
```

```python
#This cell creates the alignement object
align=Align.PairwiseAligner()
align
```

```python
# Assigning the target = the reference sequence and the query = the sequence we will align
target=wuhan_seq
query=omicron_seq
```

```python
#This cell will perform the alignement
alignements=align.align(target, query)
```

```python
#This cell gives us the best alignement scores
best=alignements[0]
best
```

```python
#This cell prints the alignment
print(best)
```

```python
# The following cell detects the substitutions and displays them clearly
aligned_target=str(best[0])
aligned_query=str(best[1])
substitutions=[]
for i, (t, q) in enumerate(zip(aligned_target, aligned_query)):
    if t!= q and t != '-' and q != '-':
        substitutions.append({'position': i, 'wuhan': t, 'omicron': q})
print(f"total substitutions found: {len(substitutions)}")
```

```python
# To get presentable results, we used pandas to put them in a clean table
### ITS IS IMPORTANT TO MENTION THAT ONLY SUBSTITUTIONS WERE COUNTED NOT INDELS
import pandas as pd

df=pd.DataFrame(substitutions)
df['mutation']=df['wuhan']+str(' ')+df['omicron']
df.head(20)
```
